# numpy.concatenate() fuction concentrate a sequence of arrays along an existing axis

```python
c= np.arange(6).reshape(2,3)
d=np.arange(6,12).reshape(2,3)

print(c)
print(d)
```

```python
np.concatenate((c,d),axis=0)
```

```python
np.concatenate((c,d),axis=1)
```

# The numpy.where() fuctions return the indices of element in an input array where the given conditions is satisfied

```python
a
```

```python
# find all indices with value greater than 50
np.where(a>50)
```

```python
#replace all values greater than 50 with 0
#np.where(condition,if true,if false)
np.where(a>50,0,a)
```

# Set Functions

```python
#np.union1d
m=np.array([1,2,3,4,5])
n=np.array([3,4,5,6,7])

np.union1d(m,n)
```

```python
np.intersect1d(m,n)
```

```python
np.setdiff1d(m,n)
```

```python
np.setxor1d(m,n)
```

```python
np.in1d(m,1)
```

```python
import numpy as np
arr = np.arange(1, 21)
```

```python
print(arr[:5])
print(arr[-5:])
print(arr[5:15])
```

```python
arr=np.arange(1,26).reshape(5,5)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
```

```python
arr=np.arange(10,60,10)
arr=arr.astype(np.float64)
print(arr)
```

```python
arr = np.arange(1, 11)
print(arr[arr%2==0])
```

```python
arr = np.array([10, 25, 3, 48, 17, 90, 6])
print(np.max(arr))
print(np.min(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
```

```python
arr = np.arange(1, 13).reshape(3, 4)
print(np.sum(arr))
print(np.sum(arr,axis=1))
print(np.sum(arr,axis=0))
```

```python
A = np.arange(1, 10).reshape(3, 3)
print(A.T)
print(A.ravel())
print(A*10)
```

```python
arr = np.arange(1, 10).reshape(3, 3)
values = np.array([10, 20, 30])
print(arr+values)
```

```python
arr = np.array([12, 5, 27, 8, 45, 19, 3, 31])
new_arr=arr[arr>20]
print(new_arr)
```

```python
marks = np.array([
    [78, 85, 91],
    [65, 72, 80],
    [90, 88, 95],
    [55, 68, 70]
])
print(np.mean(marks,axis=1))
print(np.mean(marks,axis=0))
print(np.max(marks))
print(np.argmax(np.mean(marks,axis=1)))
print(marks[marks>85])
```

axis 0 is used for columns it is used for functions like sum,max and more if we want to use these functions only on columns

axis 1 is used for rows

In NumPy, arr.mean(axis=0) gives one value per column because axis=0 points down the rows.

arr.reshpae(3,3) reshapes the array into given dimensions of row and column and it even can be of any dimension whereas ravel flatten the given array into 1d from higher dimnesions

It describes how numpy treats arrays with diffrent shapes during airthmetic operations
The smaller array is "broadcast" across the larger array so that they have compatible shapes

# Array Attributes

```python
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
```

```python
#ndim => no of dimensions

a1.ndim
a2.ndim
a3.ndim
```

```python
#shape => tell us no of rows and colums in a matrix

print(a1.shape)
print(a2.shape)
print(a3.shape)
```

```python
#size => tells no of elements

print(a1.size)
print(a2.size)
print(a3.size)
```

```python
#item size => no of bytes an item is storing in an array

print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
```

```python
#dtype => tell us the data type of elemnts in our numpy array

print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

# Array Operations

```python
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
print(a1)
print(a2)
```

```python
#scalar operation

#airthmetic
a1*2

#relational
a2>16
```

```python
#vector operations(for same shape)

#airthmetic

a1 * a2
```

# Array Functions

```python
a1=np.random.random((3,3))
a1=np.round(a1*100)
a1
```

```python
#max/min/sum/prod 
np.max(a1)
np.min(a1)
np.sum(a1)
np.prod(a1)
np.max(a1,axis=1) #0->column 1->row works with all functions
```

```python
#mean/median/std/var
np.mean(a1)
np.median(a1,axis=1)
np.std(a1,axis=1)
np.var(a1,axis=1)
```

```python
#trignometric functions
np.sin(a1)
```

```python
# dot product

a2=np.arange(12).reshape(3,4)
a3=np.arange(12,24).reshape(4,3)

np.dot(a1,a2)
```

```python
#log and exponents
np.log(a1)
np.exp(a1)
```

```python
# round/floor/ceil

np.round(np.random.random((2,3))*100)
np.floor(np.random.random((2,3))*100)
np.ceil(np.random.random((2,3))*100)
```

# Stacking

```python
# Horizontal Stacking

a4=np.arange(12).reshape(3,4)
a5=np.arange(12,24).reshape(3,4)

np.hstack((a4,a5))
```
